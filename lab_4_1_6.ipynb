Классы: Account (баланс), Client (ИНН).

Конструктор __init__ инициализируют атрибуты.

Метод __str__ возвращает форматированную строку для вывода через print().



```python
class Account:
    """Базовый класс банковского счёта."""

    def __init__(self, balance: float):
        if balance < 0:
            raise ValueError("Баланс не может быть отрицательным.")
        self.balance = balance

    def __str__(self):    # возвращает строковое представление счёта
        return f"Банковский счёт | Баланс: {self.balance:.2f} руб."


class Client:
    """Базовый класс клиента банка."""
    def __init__(self, inn: str):
        if not (inn.isdigit() and len(inn) in (10, 12)):
            raise ValueError("ИНН должен содержать 10 или 12 цифр.")
        self.inn = inn

    def __str__(self):
        return f"Клиент | ИНН: {self.inn}"



"""
Демонстрации работы классов Account и Client
"""

if __name__ == "__main__":
    # Демонстрация корректных данных
    print("Корректные объекты:")

    acc1 = Account(15000.50)
    acc2 = Account(0.0)
    client1 = Client("1234567890")
    print(acc1)
    print(acc2)
    print(client1)


    # Демонстрация обработки ошибок
    print("\nПроверка некорректных данных:")

    try:
        bad_acc = Account(-1000.0)
    except ValueError as e:
        print(f"Уважаемый пользователь! {e} Пожалуйста, введите положительное число.")

    try:
        bad_client = Client("12345")
    except ValueError as e:
        print(f"Уважаемый пользователь! {e} Проверьте длину ИНН.")

    try:
        bad_client2 = Client("12a4567890")
    except ValueError as e:
        print(f"Уважаемый пользователь! {e} Используйте только цифры.")
```
